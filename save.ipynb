```python
import cv2
import face_recognition
import pickle

cap = cv2.VideoCapture(1)

print("Press SPACE to capture teacher's face...")

while True:
    ret, frame = cap.read()
    if not ret:
        continue

    # Mirror the frame
    frame = cv2.flip(frame, 1)
    
    # Detect face
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    boxes = face_recognition.face_locations(rgb)

    # Draw rectangles around detected faces
    for (top, right, bottom, left) in boxes:
        cv2.rectangle(frame, (left, top), (right, bottom), (0, 255, 255), 2)
        cv2.putText(frame, "Press SPACE to Save", (left, top - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 255), 2)

    cv2.imshow("Capture Teacher Face", frame)

    key = cv2.waitKey(1)
    
    if key == 32 and boxes:  # SPACE key
        encodings = face_recognition.face_encodings(rgb, boxes)
        with open("teacher_face.pkl", "wb") as f:
            pickle.dump(encodings[0], f)
        print("✅ Teacher face saved!")
        break

    elif key == ord('q'):
        print("❌ Exited without saving.")
        break

cap.release()
cv2.destroyAllWindows()
```
